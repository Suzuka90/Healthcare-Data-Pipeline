```python
import matplotlib.pyplot as plt

# 1. Trova colonne categoriche (testo)
categoriche = df.select_dtypes(include=['object']).columns
print("📂 Colonne testo:", list(categoriche))

# 2. Trova colonne numeriche (numeri)
numeriche = df.select_dtypes(include=['number']).columns
print("💰 Colonne numeri:", list(numeriche))

# 3. GRAFICO 1: Top categoria (usa PRIMA disponibile)
if len(categoriche) > 0:
    top_cat = df[categoriche[0]].value_counts().head(8)
    top_cat.plot(kind='bar', color='coral', figsize=(10,6))
    plt.title(f'🏥 Top {categoriche[0]}')
    plt.xticks(rotation=45)
    plt.savefig('top_categoria.png')
    plt.show()

# 4. GRAFICO 2: Media prima colonna numerica
if len(numeriche) > 0:
    prima_num = numeriche[0]
    if len(categoriche) > 0:
        media_cat = df.groupby(categoriche[0])[prima_num].mean()
        media_cat.plot(kind='bar', color='lightblue')
        plt.title(f'💰 {prima_num} per {categoriche[0]}')
        plt.xticks(rotation=45)
        plt.savefig('costi_categoria.png')
        plt.show()

print("✅ File generati:")
print("- top_categoria.png")
print("- costi_categoria.png")
print("- Dataset:", df.shape)
```
